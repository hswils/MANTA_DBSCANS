```python
import xarray as xr
from sys import argv
import matplotlib.pyplot as plt
import numpy as np

ffile = '/pscratch/sd/h/hswils/github_repos/MANTA_DBSCANS/basecase/SCAN/work/rf_ic_toric_5/toric.nc'

rootgrp = xr.open_dataset(ffile)

Xplasma = rootgrp['Xplasma']
Zplasma = rootgrp['Zplasma']
power_absorbed_2D = rootgrp['TDPwIF']
pa_2D_firstharm = rootgrp['TdPwIH']

fig, (ax1) = plt.subplots(1,1)

#print(power_absorbed_2D.coords)

CS = ax1.contourf(Xplasma,Zplasma,pa_2D_firstharm.sel(SpecDim=0),cmap = 'cool')
ax1.set_aspect('equal')
colorbar = fig.colorbar(CS)

print(power_absorbed_2D)
```

```python
from netCDF4 import Dataset
from sys import argv
import os
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl

with plt.rc_context({
    #"text.usetex": True,
    "font.family": "serif",
    #"font.serif": ["Computer Modern Roman"],
    "font.size": 16,
    "figure.autolayout":True,
    #"legend.frameon" : False,
    "lines.linewidth" : 2.5,
    "lines.markeredgewidth" : 0.0,
    "lines.markersize" : 6.0

}):
    
    fig = plt.figure('fig_offset', figsize=(6,6),dpi = 150)
    gs = mpl.gridspec.GridSpec(2,3, figure=fig, width_ratios=[1,1,1])
    axs = {}
    axs[1] = fig.add_subplot(gs[0,:])
    axs[2] = fig.add_subplot(gs[1,0])
    #axs[3] = fig.add_subplot(gs[0,1])
    axs[4] = fig.add_subplot(gs[1,1])
    #axs[5] = fig.add_subplot(gs[0,2])
    axs[6] = fig.add_subplot(gs[1,2])

    
    ax1 = axs[1]
    ax2 = axs[2]
    #ax3 = axs[3]
    ax4 = axs[4]
    #ax5 = axs[5]
    ax6 = axs[6]

    c1 = 'magenta'
    c2 = 'dodgerblue'

    # First just find the fastran.nc file ================
    cwd = os.getcwd()
    SCAN_dir = cwd+'/SCAN'
    sims = [f for f in os.listdir(SCAN_dir) if f.startswith('simulation_') and not f.endswith('.conf') and not f.endswith('log')]


    pfus_array = []
    He3_frac_array = []
    #first column plots ----------------------------------------------------------------------------------------------------------
    sim = sims[1]
    work_dir = SCAN_dir+'/'+sim+'/work'
    
    fastran_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_tr_')]
    ffile = work_dir+'/'+fastran_dir[0]+'/fastran.nc'
    
    #toray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_ec_')]
    #toray_file = work_dir+'/'+toray_dir[0]+'/toray.nc'
    #genray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_lh_')]
    #lhfile = work_dir+'/'+genray_dir[0]+'/genray.nc'
    gfile = work_dir+'/'+fastran_dir[0]+'/g000001.00001'

    rootgrp = Dataset(ffile)
    #geqdsk = omfit_eqdsk.OMFITgeqdsk(gfile)
    #lhgrp = Dataset(lhfile)
    
#=====================================================
    rho = rootgrp['rho'][:]
    prfe = rootgrp['pe_rf'][-1,:]
    prfe_int = rootgrp['prfe'][-1]
    prfi = rootgrp['pi_rf'][-1,:]
    prfi_int = rootgrp['prfi'][-1] 
    pfuse = rootgrp['pe_fus'][-1,:]
    pfuse_int = rootgrp['pfuse'][-1]
    pfusi = rootgrp['pi_fus'][-1,:]
    pfusi_int = rootgrp['pfusi'][-1]
    pfus_array.append(5*pfuse_int+5*pfusi_int)
    nHe3 = rootgrp['nz1'][-1,:]
    ne = rootgrp['ne'][-1,:]
    He3_frac = nHe3[0]/ne[0]
    He3_frac_array.append(He3_frac)
    
    #ax5.plot(rho,pfuse,c=c1,label='electron')
    #ax5.plot(rho,pfusi,c=c2,label='ion')
    ax6.plot(rho,prfe,c=c1)
    ax6.plot(rho,prfi,c=c2)
    #ax5.set_title('$f_\\mathrm{He3} = 0.035$')

    #second column plots ----------------------------------------------------------------------------------------------------------
    sim = sims[2]
    work_dir = SCAN_dir+'/'+sim+'/work'
    
    fastran_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_tr_')]
    ffile = work_dir+'/'+fastran_dir[0]+'/fastran.nc'
    print(ffile)
    #toray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_ec_')]
    #toray_file = work_dir+'/'+toray_dir[0]+'/toray.nc'
    #genray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_lh_')]
    #lhfile = work_dir+'/'+genray_dir[0]+'/genray.nc'
    gfile = work_dir+'/'+fastran_dir[0]+'/g000001.00001'

    rootgrp = Dataset(ffile)
    #geqdsk = omfit_eqdsk.OMFITgeqdsk(gfile)
    #lhgrp = Dataset(lhfile)
    
#=====================================================
    rho = rootgrp['rho'][:]
    prfe = rootgrp['pe_rf'][-1,:]
    prfe_int = rootgrp['prfe'][-1]
    prfi = rootgrp['pi_rf'][-1,:]
    prfi_int = rootgrp['prfi'][-1] 
    pfuse = rootgrp['pe_fus'][-1,:]
    pfuse_int = rootgrp['pfuse'][-1]
    pfusi = rootgrp['pi_fus'][-1,:]
    pfusi_int = rootgrp['pfusi'][-1]
    pfus_array.append(5*pfuse_int+5*pfusi_int)
    nHe3 = rootgrp['nz1'][-1,:]
    ne = rootgrp['ne'][-1,:]
    He3_frac = nHe3[0]/ne[0]
    He3_frac_array.append(He3_frac)
    
    #ax1.plot(rho,5*pfuse,c=c1,label='electron')
    #ax1.plot(rho,5*pfusi,c=c2,label='ion')
    ax2.plot(rho,prfe,c=c1,label='electron')
    ax2.plot(rho,prfi,c=c2,label='ion')
    #ax1.set_title('$f_\\mathrm{He3} = 0.015$')

    #third column plots ----------------------------------------------------------------------------------------------------------
    sim = sims[3]
    work_dir = SCAN_dir+'/'+sim+'/work'
    
    fastran_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_tr_')]
    ffile = work_dir+'/'+fastran_dir[0]+'/fastran.nc'
    print(ffile)
    #toray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_ec_')]
    #toray_file = work_dir+'/'+toray_dir[0]+'/toray.nc'
    #genray_dir = [f for f in os.listdir(work_dir) if f.startswith('fastran_lh_')]
    #lhfile = work_dir+'/'+genray_dir[0]+'/genray.nc'
    gfile = work_dir+'/'+fastran_dir[0]+'/g000001.00001'

    rootgrp = Dataset(ffile)
    #geqdsk = omfit_eqdsk.OMFITgeqdsk(gfile)
    #lhgrp = Dataset(lhfile)
    
#=====================================================
    rho = rootgrp['rho'][:]
    prfe = rootgrp['pe_rf'][-1,:]
    prfe_int = rootgrp['prfe'][-1]
    prfi = rootgrp['pi_rf'][-1,:]
    prfi_int = rootgrp['prfi'][-1] 
    pfuse = rootgrp['pe_fus'][-1,:]
    pfuse_int = rootgrp['pfuse'][-1]
    pfusi = rootgrp['pi_fus'][-1,:]
    pfusi_int = rootgrp['pfusi'][-1]
    pfus_array.append(5*pfuse_int+5*pfusi_int)
    nHe3 = rootgrp['nz1'][-1,:]
    ne = rootgrp['ne'][-1,:]
    He3_frac = nHe3[0]/ne[0]
    He3_frac_array.append(He3_frac)

    
    #ax3.plot(rho,5*pfuse,c=c1)
    #ax3.plot(rho,5*pfusi,c=c2)
    ax4.plot(rho,prfe,c=c1)
    ax4.plot(rho,prfi,c=c2)
    #ax3.set_title('$f_\\mathrm{He3} = 0.02$')
    

    ax1.scatter(He3_frac_array,pfus_array)
    ax1.set_xlim(0.1,0.35)
    ax1.set_xlabel('$f_\\mathrm{He3}$')
    
    #ax1.set_ylim(0,5)
    ax2.set_ylim(0,4)
    #ax3.set_ylim(0,5)
    ax4.set_ylim(0,4)
    #ax5.set_ylim(0,5)
    ax6.set_ylim(0,4)
    
    
    ax1.set_ylabel('$P_\\mathrm{fus}$ (MW)')
    ax2.set_ylabel('$p_\\mathrm{rf}$ (MW/m$^3$)')
    ax2.set_xlabel('$\\rho$')
    ax4.set_xlabel('$\\rho$')  
    ax6.set_xlabel('$\\rho$') 
    fig.tight_layout()
    ax2.legend(fontsize=10)
    plt.show()
```
